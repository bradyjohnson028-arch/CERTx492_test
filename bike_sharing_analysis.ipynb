## Part 2: Python foundations
### Variables and data types
A variable gives a name to a value. Run the example, then read the printed value types:

- `str`: text.
- `int`: a whole number.
- `float`: a decimal number.

The rental counts below come from January 1, 2011. `fraction_example` is a separate practice value.


**Run this example — no edits needed:** Run the next code cell and keep its output visible.


```python
sample_date = "2011-01-01"
casual_rentals = 331
registered_rentals = 654
fraction_example = 0.25

print("Date:", sample_date, type(sample_date))
print("Casual rentals:", casual_rentals, type(casual_rentals))
print("Registered rentals:", registered_rentals, type(registered_rentals))
print("Practice fraction:", fraction_example, type(fraction_example))
```

### Calculate a total and a percentage
`+` adds values, `/` divides, and `*` multiplies. `print()` shows a result. `round(value, 2)` displays a number rounded to two decimal places.

Before running the next cell, predict the total rentals. Then run it and examine the percentage.


**Run this example — no edits needed:** Run the next code cell and keep its output visible.


```python
total_rentals = casual_rentals + registered_rentals
casual_percent = casual_rentals / total_rentals * 100

print("Total rentals:", total_rentals)
print("Casual rental percentage:", round(casual_percent, 2), "%")
```

## Part 3: Explore the sample DataFrame
A **DataFrame** is a table with rows and columns. Each row here describes one day.

| Column | Meaning |
| --- | --- |
| `dteday` | Date, kept as text for this activity. |

| `registered` | Number of rentals by registered users that day. |
| `cnt` | Total rentals that day: casual plus registered. |

These are counts of rentals; a rental count does not establish a count of distinct people.
